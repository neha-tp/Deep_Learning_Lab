```python
# Library imports

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay)
from sklearn.linear_model import Perceptron as SKPerceptron
from matplotlib.colors import ListedColormap

from google.colab import drive
drive.mount('/content/drive')

# Load Dataset

column_names = ["Variance", "Skewness", "Curtosis", "Entropy", "Class"]
path = "/content/drive/MyDrive/DL Lab/ex1/data_banknote_authentication.txt"
dataset = pd.read_csv(path, header=None, names=column_names)
print("BANKNOTE AUTHENTICATION DATASET")

# Display samples

print("\nFirst Five Records\n")
print(dataset.head())

# Dataset dimensions

print("\nDataset Shape")
print(dataset.shape)

print("\nNumber of Rows :", dataset.shape[0])
print("Number of Columns :", dataset.shape[1])

# Dataset Information

print("\nDataset Information\n")
dataset.info()

# Missing Values

print("\nMissing Values\n")
print(dataset.isnull().sum())

# Descriptive Statistics

print("\nDescriptive Statistics\n")
print(dataset.describe())
print()

# Class Distribution

print("\nClass Distribution\n")
print(dataset["Class"].value_counts())

sns.countplot(x="Class", data=dataset)
plt.title("Class Distribution")
plt.show()
print()

# Histograms

dataset.hist(figsize=(12,8), bins=20, edgecolor='black')
plt.suptitle("Feature Histograms")
plt.show()
print()

# Correlation Heatmap

plt.figure(figsize=(8,6))
sns.heatmap(dataset.corr(), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
print()

# Scatter Plot

plt.figure(figsize=(8,6))
plt.scatter(dataset["Variance"], dataset["Skewness"], c=dataset["Class"], cmap="coolwarm")

plt.xlabel("Variance")
plt.ylabel("Skewness")
plt.title("Scatter Plot")
plt.show()
print()

# Boxplots

plt.figure(figsize=(10,6))
sns.boxplot(data=dataset.iloc[:,:-1])
plt.title("Feature Boxplots")
plt.show()
print()

# Features

X = dataset.iloc[:, :-1].values

# Target

y = dataset.iloc[:, -1].values

# Feature Normalization

scaler = StandardScaler()
X = scaler.fit_transform(X)
print("\nNormalized Features\n")
print(X[:5])

# Train Test Split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print("\nTraining Samples :", len(X_train))
print("Testing Samples  :", len(X_test))

print("\nTraining Feature Shape :", X_train.shape)
print("Testing Feature Shape  :", X_test.shape)



class Perceptron:

    # Constructor
    def __init__(self, learning_rate=0.01, epochs=100):
        self.learning_rate = learning_rate
        self.epochs = epochs

        self.weights = None
        self.bias = None

        self.errors = []
        self.weight_history = []
        self.bias_history = []

    # Step Activation Function
    def activation(self, z):
        if z >= 0:
            return 1
        else:
            return 0

    # Train Perceptron
    def fit(self, X, y):
        samples, features = X.shape
        self.weights = np.zeros(features)
        self.bias = 0

        print("TRAINING SINGLE LAYER PERCEPTRON")

        for epoch in range(self.epochs):
            error_count = 0
            for i in range(samples):

                # Weighted sum
                z = np.dot(X[i], self.weights) + self.bias

                # Prediction
                prediction = self.activation(z)

                # Error
                error = y[i] - prediction

                if error != 0:
                    self.weights = self.weights + self.learning_rate * error * X[i]
                    self.bias = self.bias + self.learning_rate * error
                    error_count += 1

            self.errors.append(error_count)
            self.weight_history.append(self.weights.copy())
            self.bias_history.append(self.bias)

            print(f"\nEpoch {epoch+1}/{self.epochs}")
            print(f"Misclassified Samples : {error_count}")
            print("Weights :", np.round(self.weights, 4))
            print("Bias :", round(self.bias, 4))

            if error_count == 0:
                print("\nTraining Converged")
                break

        print("\nTraining Completed")

    # Predict One Sample
    def predict_sample(self, x):
        z = np.dot(x, self.weights) + self.bias
        return self.activation(z)

    # Predict Multiple Samples
    def predict(self, X):
        predictions = []
        for sample in X:
            predictions.append(self.predict_sample(sample))
        return np.array(predictions)

# Model training

learning_rate = 0.01
epochs = 100

model = Perceptron(learning_rate=learning_rate, epochs=epochs)
model.fit(X_train, y_train)

print("\n")
print("FINAL PARAMETERS")

print("Final Weights")
print(model.weights)

print("\nFinal Bias")
print(model.bias)

# Test set predictions

predictions = model.predict(X_test)
print("\nFirst 20 Predictions\n")

for i in range(20):
    print("Actual:", y_test[i], " Predicted:", predictions[i])
print()

# Performance Metrics

accuracy = accuracy_score(y_test, predictions)
precision = precision_score(y_test, predictions)
recall = recall_score(y_test, predictions)
f1 = f1_score(y_test, predictions)

print("MODEL PERFORMANCE")

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

# Confusion Matrix

cm = confusion_matrix(y_test, predictions)

print("\nConfusion Matrix")
print(cm)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
print()

# Training error vs epoch

plt.figure(figsize=(8,5))
plt.plot(range(1, len(model.errors)+1), model.errors, marker='o', linewidth=2)
plt.xlabel("Epoch")
plt.ylabel("Misclassified Samples")
plt.title("Training Error vs Epoch")
plt.grid(True)
plt.show()
print()

# Weight evolution

weights = np.array(model.weight_history)
plt.figure(figsize=(10,6))

for i in range(weights.shape[1]):
    plt.plot(weights[:,i], label=f"Weight {i+1}")

plt.xlabel("Epoch")
plt.ylabel("Weight Value")
plt.title("Weight Evolution")
plt.legend()
plt.grid(True)
plt.show()
print()

# Bias evolution

plt.figure(figsize=(8,5))
plt.plot(model.bias_history, marker='o', color='red')

plt.xlabel("Epoch")
plt.ylabel("Bias")
plt.title("Bias Evolution")
plt.grid(True)
plt.show()
print()

# Learning rate comparison

learning_rates = [0.001,0.01,0.1]
plt.figure(figsize=(10,6))

for lr in learning_rates:
    temp_model = Perceptron(learning_rate=lr, epochs=100)
    temp_model.fit(X_train,y_train)
    plt.plot(temp_model.errors, label=f"LR={lr}")

plt.xlabel("Epoch")
plt.ylabel("Misclassified Samples")
plt.title("Learning Rate Comparison")
plt.legend()
plt.grid(True)
plt.show()
print()

# Scikit-learn perceptron comparison

sk_model = SKPerceptron(max_iter=1000, eta0=0.01, random_state=42)
sk_model.fit(X_train,y_train)
sk_predictions = sk_model.predict(X_test)

print("\n")
print("SCIKIT-LEARN PERCEPTRON RESULTS")

print("Accuracy :",accuracy_score(y_test,sk_predictions))
print("Precision:",precision_score(y_test,sk_predictions))
print("Recall   :",recall_score(y_test,sk_predictions))
print("F1 Score :",f1_score(y_test,sk_predictions))

# Decision boundary using first 2 features

X_two = X[:,0:2]
X_train2,X_test2,y_train2,y_test2 = train_test_split(X_two, y, test_size=0.2, random_state=42, stratify=y)

boundary_model = Perceptron(learning_rate=0.01, epochs=100)
boundary_model.fit(X_train2,y_train2)

x_min,x_max = X_two[:,0].min()-1,X_two[:,0].max()+1
y_min,y_max = X_two[:,1].min()-1,X_two[:,1].max()+1

xx,yy=np.meshgrid(np.arange(x_min,x_max,0.02), np.arange(y_min,y_max,0.02))

grid=np.c_[xx.ravel(),yy.ravel()]

Z=boundary_model.predict(grid)
Z=Z.reshape(xx.shape)

plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap=ListedColormap(("lightblue","lightcoral")))
plt.scatter(X_two[:,0], X_two[:,1], c=y, cmap=ListedColormap(("blue","red")), edgecolors='k')

plt.xlabel("Variance")
plt.ylabel("Skewness")
plt.title("Decision Boundary")
plt.show()
print()

# Training summary

print("\n")
print("TRAINING SUMMARY")

print("Dataset Size       :",len(dataset))
print("Training Samples   :",len(X_train))
print("Testing Samples    :",len(X_test))
print("Learning Rate      :",model.learning_rate)
print("Epochs             :",len(model.errors))
print("Final Weights      :",model.weights)
print("Final Bias         :",model.bias)
print("Accuracy           :",accuracy)
print("Precision          :",precision)
print("Recall             :",recall)
print("F1 Score           :",f1)
```

```python
import numpy as np
import matplotlib.pyplot as plt

class Perceptron:

    def __init__(self, learning_rate=0.1, epochs=10):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0

    def activation(self, z):
        return 1 if z >= 0 else 0

    def predict(self, X):
        z = np.dot(X, self.weights) + self.bias
        return np.array([self.activation(i) for i in z])

    def fit(self, X, y, gate_name):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0
        update = 1

        for epoch in range(self.epochs):
            error_count = 0

            for i in range(len(X)):
                z = np.dot(X[i], self.weights) + self.bias
                prediction = self.activation(z)
                error = y[i] - prediction

                if error != 0:
                    self.weights += self.lr * error * X[i]
                    self.bias += self.lr * error

                    print()
                    print("Update", update)
                    print("Weights:", self.weights)
                    print("Bias:", self.bias)

                    plot_decision_boundary(self.weights, self.bias, X, y, gate_name, update)
                    update += 1
                    error_count += 1

            if error_count == 0:
                print("\nTraining Converged\n")
                break

def plot_decision_boundary(weights, bias, X, y, gate, update):
    plt.figure(figsize=(5,5))

    plt.scatter(X[:,0], X[:,1], c=y, s=120, cmap="bwr", edgecolors="black")
    x = np.linspace(-0.5,1.5,100)

    if abs(weights[1]) > 1e-10:
        x = np.linspace(-0.5, 1.5, 100)
        y_line = -(weights[0]*x + bias)/weights[1]
        plt.plot(x, y_line, 'k', linewidth=2)

    elif abs(weights[0]) > 1e-10:
        x_line = -bias / weights[0]
        plt.axvline(x=x_line, color='black', linewidth=2)

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)

    plt.grid(True)
    plt.title(f"{gate} Gate\nUpdate {update}")
    plt.xlabel("Input 1")
    plt.ylabel("Input 2")
    plt.show()

X_and = np.array([[0,0], [0,1], [1,0], [1,1]])
y_and = np.array([0,0,0,1])

print("AND GATE \n")

and_model = Perceptron(learning_rate=0.1, epochs=10)
and_model.fit(X_and, y_and, "AND")

print("Final Weights:", and_model.weights)
print("Final Bias:", and_model.bias)
print()

X_or = np.array([[0,0], [0,1], [1,0], [1,1]])
y_or = np.array([0,1,1,1])

print("OR GATE \n")

or_model = Perceptron(learning_rate=0.1, epochs=10)
or_model.fit(X_or, y_or, "OR")

print("Final Weights:", or_model.weights)
print("Final Bias:", or_model.bias)
print()

X_not = np.array([[0,0], [1,0]])
y_not = np.array([1,0])

print("NOT GATE\n")

not_model = Perceptron(learning_rate=0.1, epochs=10)
not_model.fit(X_not, y_not, "NOT")

print("Final Weights:", not_model.weights)
print("Final Bias:", not_model.bias)
print()
```
