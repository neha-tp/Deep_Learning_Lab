```python
import os
import json
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import (Input, Conv2D, AveragePooling2D, MaxPooling2D,
                                      Flatten, Dense, Dropout, GlobalAveragePooling2D,
                                      Concatenate)
from tensorflow.keras.applications import VGG16, ResNet50
from tensorflow.keras.optimizers import Adam, SGD
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, confusion_matrix, classification_report)

NUM_CLASSES = 10
EPOCHS_MAIN = 5
EPOCHS_FINETUNE = 5
EPOCHS_HP = 1
BATCH_SIZE = 32
class_names = ['Airplane', 'Automobile', 'Bird', 'Cat', 'Deer',
               'Dog', 'Frog', 'Horse', 'Ship', 'Truck']

CKPT_DIR = "checkpoints"
os.makedirs(CKPT_DIR, exist_ok=True)
HP_CSV = os.path.join(CKPT_DIR, "hyperparameter_study_results.csv")

# Task 1: Dataset Preparation
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
y_train_cat = to_categorical(y_train, NUM_CLASSES)
y_test_cat = to_categorical(y_test, NUM_CLASSES)
y_test_true = y_test.flatten()

print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)

sample_img_path = os.path.join(CKPT_DIR, "_sample_images_done.flag")
if not os.path.exists(sample_img_path):
    plt.figure(figsize=(10, 4))
    for i in range(10):
        plt.subplot(2, 5, i + 1)
        plt.imshow(x_train[i])
        plt.title(class_names[y_train[i][0]])
        plt.axis('off')
    plt.suptitle("Sample CIFAR-10 Images")
    plt.tight_layout()
    plt.savefig("sample_images.png")
    plt.show()
    open(sample_img_path, "w").close()

# Model builders

def build_lenet5():
    return Sequential([
        Input(shape=(32, 32, 3)),
        Conv2D(6, (5, 5), activation='relu', padding='same'),
        AveragePooling2D(pool_size=(2, 2)),
        Conv2D(16, (5, 5), activation='relu'),
        AveragePooling2D(pool_size=(2, 2)),
        Flatten(),
        Dense(120, activation='relu'),
        Dense(84, activation='relu'),
        Dense(NUM_CLASSES, activation='softmax')
    ], name="LeNet5")


def build_alexnet():
    return Sequential([
        Input(shape=(32, 32, 3)),
        Conv2D(96, (3, 3), padding='same', activation='relu'),
        MaxPooling2D((2, 2)),
        Conv2D(256, (3, 3), padding='same', activation='relu'),
        MaxPooling2D((2, 2)),
        Conv2D(384, (3, 3), padding='same', activation='relu'),
        Conv2D(384, (3, 3), padding='same', activation='relu'),
        Conv2D(256, (3, 3), padding='same', activation='relu'),
        MaxPooling2D((2, 2)),
        Flatten(),
        Dense(1024, activation='relu'),
        Dropout(0.5),
        Dense(1024, activation='relu'),
        Dropout(0.5),
        Dense(NUM_CLASSES, activation='softmax')
    ], name="AlexNet")


def inception_module(x, f1, f3r, f3, f5r, f5, fp):
    c1 = Conv2D(f1, (1, 1), padding='same', activation='relu')(x)
    c3 = Conv2D(f3r, (1, 1), padding='same', activation='relu')(x)
    c3 = Conv2D(f3, (3, 3), padding='same', activation='relu')(c3)
    c5 = Conv2D(f5r, (1, 1), padding='same', activation='relu')(x)
    c5 = Conv2D(f5, (5, 5), padding='same', activation='relu')(c5)
    p = MaxPooling2D((3, 3), strides=1, padding='same')(x)
    p = Conv2D(fp, (1, 1), padding='same', activation='relu')(p)
    return Concatenate()([c1, c3, c5, p])


def build_googlenet():
    inp = Input(shape=(32, 32, 3))
    x = Conv2D(64, (3, 3), padding='same', activation='relu')(inp)
    x = MaxPooling2D((2, 2))(x)
    x = inception_module(x, 64, 96, 128, 16, 32, 32)
    x = inception_module(x, 128, 128, 192, 32, 96, 64)
    x = MaxPooling2D((2, 2))(x)
    x = GlobalAveragePooling2D()(x)
    x = Dropout(0.4)(x)
    out = Dense(NUM_CLASSES, activation='softmax')(x)
    return Model(inp, out, name="GoogleNet")


def build_vgg16_transfer():
    base = VGG16(weights='imagenet', include_top=False, input_shape=(32, 32, 3))
    base.trainable = False
    x = base.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    out = Dense(NUM_CLASSES, activation='softmax')(x)
    return Model(base.input, out, name="VGG16"), base


def build_resnet50_transfer():
    base = ResNet50(weights='imagenet', include_top=False, input_shape=(32, 32, 3))
    base.trainable = False
    x = base.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation='relu')(x)
    out = Dense(NUM_CLASSES, activation='softmax')(x)
    return Model(base.input, out, name="ResNet50"), base


# Resumable train helper

def train_or_resume(model, tag, optimizer, epochs, batch_size=BATCH_SIZE, verbose=1):
    """Trains `model` and checkpoints it under `tag`. If a checkpoint for `tag`
    already exists, loads it instead of retraining."""
    weights_path = os.path.join(CKPT_DIR, f"{tag}.weights.h5")
    hist_path = os.path.join(CKPT_DIR, f"{tag}_history.json")
    meta_path = os.path.join(CKPT_DIR, f"{tag}_meta.json")

    model.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])

    if os.path.exists(weights_path) and os.path.exists(hist_path):
        print(f"[Resume] '{tag}' already trained — loading checkpoint, skipping training.")
        model.load_weights(weights_path)
        with open(hist_path) as f:
            hist_dict = json.load(f)
        with open(meta_path) as f:
            elapsed = json.load(f)["time_seconds"]

        class _FakeHistory:
            pass
        fake_hist = _FakeHistory()
        fake_hist.history = hist_dict
        return fake_hist, elapsed

    print(f"[Train] '{tag}' — no checkpoint found, training now.")
    start = time.time()
    hist = model.fit(x_train, y_train_cat, validation_data=(x_test, y_test_cat),
                      epochs=epochs, batch_size=batch_size, verbose=verbose)
    elapsed = time.time() - start

    model.save_weights(weights_path)
    with open(hist_path, "w") as f:
        json.dump(hist.history, f)
    with open(meta_path, "w") as f:
        json.dump({"time_seconds": elapsed}, f)

    return hist, elapsed


def evaluate_model(model):
    y_pred = np.argmax(model.predict(x_test, verbose=0), axis=1)
    acc = accuracy_score(y_test_true, y_pred)
    prec = precision_score(y_test_true, y_pred, average='macro', zero_division=0)
    rec = recall_score(y_test_true, y_pred, average='macro', zero_division=0)
    f1 = f1_score(y_test_true, y_pred, average='macro', zero_division=0)
    return acc, prec, rec, f1, y_pred


def plot_history(hist, title, filename):
    plt.figure()
    plt.plot(hist.history['accuracy'], label='Training Accuracy')
    plt.plot(hist.history['val_accuracy'], label='Validation Accuracy')
    plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.title(title + ' - Accuracy'); plt.legend()
    plt.savefig(filename + "_accuracy.png"); plt.show()

    plt.figure()
    plt.plot(hist.history['loss'], label='Training Loss')
    plt.plot(hist.history['val_loss'], label='Validation Loss')
    plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.title(title + ' - Loss'); plt.legend()
    plt.savefig(filename + "_loss.png"); plt.show()


# Task 2 & 3: VGG16 (frozen base)
vgg_model, vgg_base = build_vgg16_transfer()
vgg_model.summary()
vgg_hist, vgg_time = train_or_resume(vgg_model, "vgg16_frozen", Adam(learning_rate=0.001), EPOCHS_MAIN)
plot_history(vgg_hist, "VGG16 (Frozen Base)", "vgg16_frozen")
vgg_acc_frozen, *_ = evaluate_model(vgg_model)

# Task 4: Fine-tune VGG16
vgg_base.trainable = True
for layer in vgg_base.layers[:-4]:
    layer.trainable = False

vgg_ft_hist, vgg_ft_time = train_or_resume(vgg_model, "vgg16_finetuned", Adam(learning_rate=0.0001), EPOCHS_FINETUNE)
plot_history(vgg_ft_hist, "VGG16 (Fine Tuned)", "vgg16_finetuned")
vgg_acc, vgg_prec, vgg_rec, vgg_f1, vgg_pred = evaluate_model(vgg_model)
vgg_params = vgg_model.count_params()
vgg_total_time = vgg_time + vgg_ft_time

print(f"\nVGG16 - Accuracy before fine-tuning: {vgg_acc_frozen:.4f}")
print(f"VGG16 - Accuracy after fine-tuning : {vgg_acc:.4f}")

# Task 5: Evaluation for VGG16 - Confusion Matrix
print("\nVGG16 Classification Report:\n",
      classification_report(y_test_true, vgg_pred, target_names=class_names))

cm = confusion_matrix(y_test_true, vgg_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label'); plt.ylabel('True Label'); plt.title('Confusion Matrix - VGG16')
plt.tight_layout(); plt.savefig("confusion_matrix_vgg16.png"); plt.show()

# Additional Exercise 2: ResNet50 (transfer learning + fine-tune)
resnet_model, resnet_base = build_resnet50_transfer()
resnet_hist, resnet_time = train_or_resume(resnet_model, "resnet50_frozen", Adam(learning_rate=0.001), EPOCHS_MAIN)
plot_history(resnet_hist, "ResNet50 (Frozen Base)", "resnet50_frozen")
resnet_acc_frozen, *_ = evaluate_model(resnet_model)

resnet_base.trainable = True
for layer in resnet_base.layers[:-10]:
    layer.trainable = False

resnet_ft_hist, resnet_ft_time = train_or_resume(resnet_model, "resnet50_finetuned", Adam(learning_rate=0.0001), EPOCHS_FINETUNE)
plot_history(resnet_ft_hist, "ResNet50 (Fine Tuned)", "resnet50_finetuned")
resnet_acc, resnet_prec, resnet_rec, resnet_f1, resnet_pred = evaluate_model(resnet_model)
resnet_params = resnet_model.count_params()
resnet_total_time = resnet_time + resnet_ft_time

print(f"\nResNet50 - Accuracy before fine-tuning: {resnet_acc_frozen:.4f}")
print(f"ResNet50 - Accuracy after fine-tuning : {resnet_acc:.4f}")

# LeNet-5, AlexNet, GoogleNet (trained from scratch on CIFAR-10)
lenet_model = build_lenet5()
lenet_hist, lenet_time = train_or_resume(lenet_model, "lenet5", Adam(learning_rate=0.001), EPOCHS_MAIN)
plot_history(lenet_hist, "LeNet-5", "lenet5")
lenet_acc, lenet_prec, lenet_rec, lenet_f1, lenet_pred = evaluate_model(lenet_model)
lenet_params = lenet_model.count_params()

alexnet_model = build_alexnet()
alexnet_hist, alexnet_time = train_or_resume(alexnet_model, "alexnet", Adam(learning_rate=0.001), EPOCHS_MAIN)
plot_history(alexnet_hist, "AlexNet", "alexnet")
alexnet_acc, alexnet_prec, alexnet_rec, alexnet_f1, alexnet_pred = evaluate_model(alexnet_model)
alexnet_params = alexnet_model.count_params()

googlenet_model = build_googlenet()
googlenet_hist, googlenet_time = train_or_resume(googlenet_model, "googlenet", Adam(learning_rate=0.001), EPOCHS_MAIN)
plot_history(googlenet_hist, "GoogleNet", "googlenet")
googlenet_acc, googlenet_prec, googlenet_rec, googlenet_f1, googlenet_pred = evaluate_model(googlenet_model)
googlenet_params = googlenet_model.count_params()

# Section 18.2: Comparison Table across all 5 architectures
comparison_table = pd.DataFrame([
    {"Model": "LeNet-5",   "Parameters": lenet_params,    "Accuracy (%)": round(lenet_acc * 100, 2),    "Training Time (s)": round(lenet_time, 2)},
    {"Model": "AlexNet",   "Parameters": alexnet_params,  "Accuracy (%)": round(alexnet_acc * 100, 2),  "Training Time (s)": round(alexnet_time, 2)},
    {"Model": "VGG16",     "Parameters": vgg_params,      "Accuracy (%)": round(vgg_acc * 100, 2),      "Training Time (s)": round(vgg_total_time, 2)},
    {"Model": "GoogleNet", "Parameters": googlenet_params,"Accuracy (%)": round(googlenet_acc * 100, 2),"Training Time (s)": round(googlenet_time, 2)},
    {"Model": "ResNet50",  "Parameters": resnet_params,   "Accuracy (%)": round(resnet_acc * 100, 2),   "Training Time (s)": round(resnet_total_time, 2)},
])
print("\nSection 18.2 - Comparison of CNN Architectures:\n", comparison_table.to_string(index=False))
comparison_table.to_csv("comparison_table_18_2.csv", index=False)

print("\nAdditional Exercise 6 - LeNet vs AlexNet vs ResNet:")
print(comparison_table[comparison_table["Model"].isin(["LeNet-5", "AlexNet", "ResNet50"])].to_string(index=False))

# Section 16: Hyperparameter Study (resumable, row-by-row CSV)
# Additional Exercise 3 (Adam vs SGD) + Exercise 4 (frozen vs partial)
learning_rates = [0.001, 0.0001]
batch_sizes = [16, 32, 64]
optimizers = {"Adam": Adam, "SGD": SGD}
dense_units_options = [128, 256]
frozen_options = ["All", "Partial"]

if os.path.exists(HP_CSV):
    hp_table = pd.read_csv(HP_CSV)
    done_keys = set(zip(hp_table["Learning Rate"], hp_table["Batch Size"], hp_table["Optimizer"],
                         hp_table["Dense Units"], hp_table["Frozen Layers"]))
else:
    hp_table = pd.DataFrame(columns=["Learning Rate", "Batch Size", "Optimizer",
                                      "Dense Units", "Frozen Layers", "Validation Accuracy"])
    hp_table.to_csv(HP_CSV, index=False)
    done_keys = set()

for lr in learning_rates:
    for bs in batch_sizes:
        for opt_name, opt_fn in optimizers.items():
            for dense_units in dense_units_options:
                for frozen in frozen_options:
                    key = (lr, bs, opt_name, dense_units, frozen)
                    if key in done_keys:
                        print(f"[Resume] Skipping already-done combo: {key}")
                        continue

                    base = VGG16(weights='imagenet', include_top=False, input_shape=(32, 32, 3))
                    if frozen == "All":
                        base.trainable = False
                    else:
                        base.trainable = True
                        for layer in base.layers[:-4]:
                            layer.trainable = False

                    h = base.output
                    h = GlobalAveragePooling2D()(h)
                    h = Dense(dense_units, activation='relu')(h)
                    out = Dense(NUM_CLASSES, activation='softmax')(h)
                    temp_model = Model(base.input, out)
                    temp_model.compile(optimizer=opt_fn(learning_rate=lr),
                                        loss='categorical_crossentropy', metrics=['accuracy'])

                    hist = temp_model.fit(x_train, y_train_cat, validation_data=(x_test, y_test_cat),
                                           epochs=EPOCHS_HP, batch_size=bs, verbose=0)
                    val_acc = round(hist.history['val_accuracy'][-1], 4)

                    row = pd.DataFrame([{
                        "Learning Rate": lr, "Batch Size": bs, "Optimizer": opt_name,
                        "Dense Units": dense_units, "Frozen Layers": frozen,
                        "Validation Accuracy": val_acc
                    }])
                    row.to_csv(HP_CSV, mode='a', header=False, index=False)  # save immediately
                    hp_table = pd.concat([hp_table, row], ignore_index=True)

                    print(f"LR={lr}, BS={bs}, Opt={opt_name}, Dense={dense_units}, "
                          f"Frozen={frozen} -> Val Acc={val_acc:.4f}")

print("\nSection 16 - Hyperparameter Study Results:\n", hp_table.to_string(index=False))

print("\nAdditional Exercise 3 - Adam vs SGD (mean validation accuracy):")
print(hp_table.groupby("Optimizer")["Validation Accuracy"].mean())

print("\nAdditional Exercise 4 - Frozen vs Partial fine-tuning (mean validation accuracy):")
print(hp_table.groupby("Frozen Layers")["Validation Accuracy"].mean())
print(f"\n(Direct VGG16 result) Frozen base accuracy: {vgg_acc_frozen:.4f} | Fine-tuned accuracy: {vgg_acc:.4f}")

# Additional Exercise 5: Precision, Recall, F1-score for all models
metrics_table = pd.DataFrame([
    {"Model": "LeNet-5",   "Precision": round(lenet_prec, 4),    "Recall": round(lenet_rec, 4),    "F1-score": round(lenet_f1, 4)},
    {"Model": "AlexNet",   "Precision": round(alexnet_prec, 4),  "Recall": round(alexnet_rec, 4),  "F1-score": round(alexnet_f1, 4)},
    {"Model": "VGG16",     "Precision": round(vgg_prec, 4),      "Recall": round(vgg_rec, 4),      "F1-score": round(vgg_f1, 4)},
    {"Model": "GoogleNet", "Precision": round(googlenet_prec, 4),"Recall": round(googlenet_rec, 4),"F1-score": round(googlenet_f1, 4)},
    {"Model": "ResNet50",  "Precision": round(resnet_prec, 4),   "Recall": round(resnet_rec, 4),   "F1-score": round(resnet_f1, 4)},
])
print("\nAdditional Exercise 5 - Precision/Recall/F1 for all models:\n", metrics_table.to_string(index=False))
metrics_table.to_csv("precision_recall_f1_table.csv", index=False)

print("\nAll tasks and additional exercises completed. CSVs and plots saved to the working directory.")
```
