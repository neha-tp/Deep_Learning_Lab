```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import time

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam, SGD, RMSprop

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

from sklearn.model_selection import RandomizedSearchCV

from scikeras.wrappers import KerasClassifier

# Load Dataset

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

print("Training Images :", X_train.shape)
print("Training Labels :", y_train.shape)

print("Testing Images :", X_test.shape)
print("Testing Labels :", y_test.shape)

class_names = ['T-shirt','Trouser','Pullover','Dress', 'Coat','Sandal','Shirt','Sneaker', 'Bag','Ankle Boot']

plt.figure(figsize=(12,6))

for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(class_names[y_train[i]])
    plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8,5))
sns.countplot(x=y_train)
plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()

print("Before Flattening")

print(X_train.shape)
print(X_test.shape)

# Flatten

X_train_flat = X_train.reshape(-1, 784)
X_test_flat = X_test.reshape(-1, 784)

# Normalize

X_train_flat = X_train_flat.astype('float32')/255.0
X_test_flat = X_test_flat.astype('float32')/255.0

# One Hot Encoding

y_train_cat = to_categorical(y_train,10)
y_test_cat = to_categorical(y_test,10)

print()

print("After Flattening")

print(X_train_flat.shape)
print(X_test_flat.shape)

print(y_train_cat.shape)
print(y_test_cat.shape)

baseline_model = Sequential([Input(shape=(784,)), Dense(128, activation='relu'), Dense(64, activation='relu'), Dense(10, activation='softmax')])
baseline_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
baseline_model.summary()

start = time.time()
history = baseline_model.fit(X_train_flat, y_train_cat, epochs=20, batch_size=32, validation_split=0.2, verbose=1)
baseline_training_time = time.time()-start

pred = baseline_model.predict(X_test_flat)
pred_classes = np.argmax(pred,axis=1)

accuracy = accuracy_score(y_test,pred_classes)
precision = precision_score(y_test,pred_classes,average='weighted')
recall = recall_score(y_test,pred_classes,average='weighted')
f1 = f1_score(y_test,pred_classes,average='weighted')

print("Accuracy :",accuracy)
print("Precision :",precision)
print("Recall :",recall)
print("F1 Score :",f1)

cm = confusion_matrix(y_test,pred_classes)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

print(classification_report(y_test, pred_classes, target_names=class_names))

plt.figure(figsize=(8,5))
plt.plot(history.history['accuracy'])
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()

plt.figure(figsize=(8,5))
plt.plot(history.history['val_accuracy'])
plt.title("Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()

plt.figure(figsize=(8,5))
plt.plot(history.history['loss'])
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

plt.figure(figsize=(8,5))
plt.plot(history.history['val_loss'])
plt.title("Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

def create_model(hidden_layers=2, neurons=128, learning_rate=0.001, optimizer='adam', activation='relu', dropout=0.0):
    model = Sequential()
    model.add(Input(shape=(784,)))
    model.add(Dense(neurons, activation=activation))
    model.add(Dropout(dropout))

    for i in range(hidden_layers-1):
        model.add(Dense(neurons, activation=activation))
        model.add(Dropout(dropout))

    model.add(Dense(10, activation='softmax'))

    if optimizer=="adam":
        opt=Adam(learning_rate=learning_rate)

    elif optimizer=="sgd":
        opt=SGD(learning_rate=learning_rate)

    else:
        opt=RMSprop(learning_rate=learning_rate)

    model.compile( optimizer=opt, loss='categorical_crossentropy', metrics=['accuracy'])
    return model

classifier = KerasClassifier(model=create_model, verbose=0)
param_grid = {"model__hidden_layers":[1,2,3], "model__neurons":[32,64,128,256], "model__learning_rate":[0.1,0.01,0.001], "model__optimizer":["adam","sgd","rmsprop"], "model__activation":["relu","tanh","sigmoid"], "model__dropout":[0.0,0.2,0.5], "batch_size":[16,32,64,128], "epochs":[10,20,30]}

random_search = RandomizedSearchCV(estimator=classifier, param_distributions=param_grid, n_iter=10, cv=5, scoring='accuracy', verbose=2, random_state=42)
random_search.fit(X_train_flat,y_train_cat)

print("Best Parameters")
print(random_search.best_params_)

print()

print("Best CV Accuracy")
print(random_search.best_score_)

best_model = random_search.best_estimator_
start = time.time()
best_model.fit(X_train_flat,y_train_cat)
optimized_training_time = time.time()-start

pred_best = best_model.predict(X_test_flat)

if pred_best.ndim == 2:
    pred_best = np.argmax(pred_best, axis=1)

print("\nOptimized Classification Report")
print(classification_report(y_test, pred_best, target_names=class_names))

accuracy_best = accuracy_score(y_test,pred_best)
precision_best = precision_score(y_test, pred_best, average='weighted')
recall_best = recall_score(y_test, pred_best, average='weighted')
f1_best = f1_score(y_test, pred_best, average='weighted')

print("Optimized Accuracy :",accuracy_best)
print("Optimized Precision :",precision_best)
print("Optimized Recall :",recall_best)
print("Optimized F1 :",f1_best)

cm = confusion_matrix(y_test,pred_best)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, cmap='Greens', fmt='d')
plt.title("Optimized Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

results = pd.DataFrame(random_search.cv_results_)

plt.figure(figsize=(8,5))
plt.plot(results['mean_test_score'])
plt.xlabel("Iteration")
plt.ylabel("CV Accuracy")
plt.title("Hyperparameter Search Results")
plt.show()

plt.figure(figsize=(6,5))
models=["Baseline","Optimized"]
scores=[accuracy,accuracy_best]
plt.bar(models,scores,color=['blue','green'])
plt.ylabel("Accuracy")
plt.title("Best Model Accuracy Comparison")
plt.show()

results = pd.DataFrame({'Metric':['Accuracy','Precision','Recall','F1 Score','Training Time'], 'Baseline':[accuracy, precision, recall, f1, baseline_training_time], 'Optimized':[accuracy_best, precision_best, recall_best, f1_best, optimized_training_time]})
print(results)
```

```python
import numpy as np
import matplotlib.pyplot as plt


class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=10):
        self.lr = learning_rate
        self.epochs = epochs

        self.weights = None
        self.bias = 0.0

        self.update_count = 0
        self.error_history = []

    def activation(self, z):
        return 1 if z >= 0 else 0

    def predict(self, X):
        z = np.dot(X, self.weights) + self.bias
        return np.array([self.activation(value) for value in z])

    def fit(self, X, y, gate_name):
        self.weights = np.zeros(X.shape[1], dtype=float)
        self.bias = 0.0

        for epoch in range(1, self.epochs + 1):
            print(f"Epoch {epoch}")
            error_count = 0

            for i in range(len(X)):
                z = np.dot(X[i], self.weights) + self.bias
                prediction = self.activation(z)
                error = y[i] - prediction

                print(f"Input: {X[i]}, " f"Target: {y[i]}, " f"Prediction: {prediction}, " f"Error: {error}")

                if error != 0:
                    old_weights = self.weights.copy()
                    old_bias = self.bias

                    self.weights += self.lr * error * X[i]
                    self.bias += self.lr * error

                    self.update_count += 1
                    error_count += 1

                    print(f"\nUpdate {self.update_count}")
                    print("Old weights:", old_weights)
                    print("New weights:", self.weights)
                    print("Old bias:", old_bias)
                    print("New bias:", self.bias)

                    plot_decision_boundary(weights=self.weights, bias=self.bias, X=X, y=y, gate=gate_name, epoch=epoch, update=self.update_count)

            self.error_history.append(error_count)

            print(f"\nErrors in epoch {epoch}: {error_count}")
            print("Weights at end of epoch:", self.weights)
            print("Bias at end of epoch:", self.bias)

            if error_count == 0:
                print("\nTraining converged.")
                return

        print("\nTraining did not converge.")
        print("Reason: XOR is not linearly separable.")


def plot_decision_boundary(weights, bias, X, y, gate, epoch, update):
    plt.figure(figsize=(5, 5))

    class_0 = X[y == 0]
    plt.scatter(class_0[:, 0], class_0[:, 1], color="blue", s=140, edgecolors="black", label="Output 0", zorder=3)
    class_1 = X[y == 1]
    plt.scatter(class_1[:, 0], class_1[:, 1], color="red", s=140, edgecolors="black", label="Output 1", zorder=3)

    x_values = np.linspace(-0.5, 1.5, 100)

    if abs(weights[1]) > 1e-10:
        y_values = -(weights[0] * x_values + bias) / weights[1]
        plt.plot(x_values, y_values, color="black", linewidth=2, label="Decision boundary", zorder=2)

    elif abs(weights[0]) > 1e-10:
        x_boundary = -bias / weights[0]
        plt.axvline(x=x_boundary, color="black", linewidth=2, label="Decision boundary", zorder=2)

    else:
        plt.text(0.5, 0.5, "No unique decision boundary\n(weights are zero)", ha="center", va="center", fontsize=10)

    plt.xlim(-0.5, 1.5)
    plt.ylim(-0.5, 1.5)

    plt.xticks([0, 1])
    plt.yticks([0, 1])

    plt.xlabel("Input 1")
    plt.ylabel("Input 2")

    plt.title(f"{gate} Gate\n" f"Epoch {epoch}, Update {update}\n"f"Weights = {weights}, Bias = {bias:.1f}")

    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()


def plot_error_history(error_history):
    epochs = range(1, len(error_history) + 1)

    plt.figure(figsize=(7, 4))
    plt.plot(epochs, error_history, marker="o", linewidth=2)

    plt.xlabel("Epoch")
    plt.ylabel("Number of weight updates")
    plt.title("XOR Perceptron: Errors per Epoch")
    plt.xticks(list(epochs))
    plt.grid(True)
    plt.tight_layout()
    plt.show()

X_xor = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_xor = np.array([0, 1, 1, 0])
print("XOR GATE")

xor_model = Perceptron(learning_rate=0.1, epochs=10)
xor_model.fit(X=X_xor, y=y_xor, gate_name="XOR")

print("\nFinal weights:", xor_model.weights)
print("Final bias:", xor_model.bias)

final_predictions = xor_model.predict(X_xor)

print("\nFinal predictions:", final_predictions)
print("Actual outputs:   ", y_xor)

plot_error_history(xor_model.error_history)
```

```python
import numpy as np
import matplotlib.pyplot as plt

class MLP:

    def __init__(self, learning_rate=1.0, epochs=10000):
        self.lr = learning_rate
        self.epochs = epochs
        np.random.seed(1)
        self.W1 = np.random.randn(2, 4)
        self.b1 = np.zeros((1, 4))
        self.W2 = np.random.randn(4, 1)
        self.b2 = np.zeros((1, 1))

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def forward(self, X):
        self.h = np.tanh(np.dot(X, self.W1) + self.b1)
        self.output = self.sigmoid(np.dot(self.h, self.W2) + self.b2)
        return self.output

    def predict(self, X):
        return (self.forward(X) >= 0.5).astype(int)

    def fit(self, X, y):
        n = len(X)

        for epoch in range(1, self.epochs + 1):
            output = self.forward(X)
            error = output - y

            dW2 = np.dot(self.h.T, error) / n
            db2 = np.mean(error, axis=0, keepdims=True)
            hidden_error = np.dot(error, self.W2.T) * (1 - self.h ** 2)
            dW1 = np.dot(X.T, hidden_error) / n
            db1 = np.mean(hidden_error, axis=0, keepdims=True)

            self.W2 -= self.lr * dW2
            self.b2 -= self.lr * db2
            self.W1 -= self.lr * dW1
            self.b1 -= self.lr * db1

            print("\nUpdate", epoch)
            print("W1:", self.W1)
            print("W2:", self.W2)

            plot_boundary(self, X, y, epoch)

            if np.array_equal(self.predict(X), y):
                print("\nTraining converged at update", epoch)
                break


def plot_boundary(model, X, y, update):
    x1 = np.linspace(-0.5, 1.5, 200)
    x2 = np.linspace(-0.5, 1.5, 200)
    xx, yy = np.meshgrid(x1, x2)
    grid = np.c_[xx.ravel(), yy.ravel()]
    output = model.forward(grid).reshape(xx.shape)

    plt.figure(figsize=(5, 5))
    plt.contourf(xx, yy, output, levels=[0, 0.5, 1], cmap="bwr", alpha=0.3)
    plt.contour(xx, yy, output, levels=[0.5], colors="black")
    plt.scatter(X[:, 0], X[:, 1], c=y.ravel(), cmap="bwr", s=120, edgecolors="black")
    plt.xlim(-0.5, 1.5)
    plt.ylim(-0.5, 1.5)
    plt.xlabel("Input 1")
    plt.ylabel("Input 2")
    plt.title(f"XOR MLP\nUpdate {update}")
    plt.grid(True)
    plt.show()


X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([[0], [1], [1], [0]])

model = MLP(learning_rate=1.0, epochs=10000)
model.fit(X, y)

print("\nFinal predictions:", model.predict(X).ravel())
print("Expected outputs:", y.ravel())
```
